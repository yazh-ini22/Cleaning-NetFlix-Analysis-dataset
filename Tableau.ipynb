```python
df['director'] = df['director'].fillna('Unknown')
```

```python
df['cast'] = df['cast'].fillna('Unknown')
```

```python
df['country'] = df['country'].fillna('Unknown')
```

```python
df['rating'] = df['rating'].fillna('Unknown')
```

```python
df['date_added'] = df['date_added'].str.strip()
```

```python
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
```

```python
df['date_added'].isnull().sum()
```

```python
df['duration_value'] = df['duration'].str.extract(r'(\d+)')
df['duration_value'] = pd.to_numeric(df['duration_value'], errors='coerce')
```

```python
df['movie_duration_min'] = np.where(
    df['type'] == 'Movie',
    df['duration_value'],
    np.nan
)
```

```python
df['seasons'] = np.where(
    df['type'] == 'TV Show',
    df['duration_value'],
    np.nan
)
```

```python
df.drop(columns=['duration_value'], inplace=True)
```

```python
text_columns = [
    'type',
    'title',
    'director',
    'cast',
    'country',
    'rating',
    'listed_in',
    'description'
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
# Create a copy of the dataset
genre_df = df.copy()

# Convert listed_in to text and handle missing values
genre_df['listed_in'] = genre_df['listed_in'].fillna('').astype(str)

# Split genres by comma
genre_df['listed_in'] = genre_df['listed_in'].str.split(',')

# Create a separate row for each genre
genre_df = genre_df.explode('listed_in')

# Remove extra spaces
genre_df['listed_in'] = genre_df['listed_in'].str.strip()

# Remove empty genre values
genre_df = genre_df[genre_df['listed_in'] != '']

# Check the result
print(genre_df[['title', 'listed_in']].head(20))
```

```python
# Remove brackets and quotes from genre names
genre_df['listed_in'] = (
    genre_df['listed_in']
    .str.replace('[', '', regex=False)
    .str.replace(']', '', regex=False)
    .str.replace("'", '', regex=False)
    .str.strip()
)

# Remove empty values
genre_df = genre_df[genre_df['listed_in'] != '']

# Check the result
print(genre_df[['title', 'listed_in']].head(20))
```

```python
genre_df.to_csv('/content/netflix_genre_analysis.csv', index=False)

print("✅ Genre analysis file created successfully!")
```
